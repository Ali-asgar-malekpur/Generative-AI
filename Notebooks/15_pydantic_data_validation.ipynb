```python
from pydantic import BaseModel, Field
from typing import List, Annotated

class User(BaseModel):
    name: str
    age: int

ali_user = User(name='Ali', age = 25) #age = '25' would throw an error
print(ali_user)

class Product(BaseModel):
    title:str = Field(max_length=100)
    desc:str = Field(min_length=20)
    price: int = Field(gt=0, le=100) #greater than, less then

product_1 = Product(title='Samsung S23', desc = 'This is amazing mobile form samsung!', price=20)
print(product_1) #Would fail since desc should have 10 chars atleast

class Order(BaseModel):
    id: str = Field(min_length=6)
    product:Product #Here we added a class in type

order = Order(id = 5, product=product_1)
print(order.product.title)
```

```python
class Carts(BaseModel):
    id: str = Field(min_length=6)
    items: List[Product] #Passing list of product that will contain product objects
    comment: str = Field(default='')

cart = Carts(id='asdfwqet', items=[product_1])

for product in cart.items:
    print(product.title)

#Annotated:
commonvalidator = Annotated(list[Product], 'This is common validator')
#Now instead of List[Product], simply mention commonvalidator (Acts as a variable)
```
