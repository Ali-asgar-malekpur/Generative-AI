```python
#Prompt Templates: Dynamic Prompts
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOpenAI(model = 'gpt-4o')
dynamic_format_prompts = ChatPromptTemplate.from_messages([
    {'role':'system','content': 'You are a translator, translate in {language}'},
    {'role':'user','content': '{query}'}
])
final_prompts = dynamic_format_prompts.format_messages(language = 'Hindi',
        query = 'I love Python and JavaScript')

response = llm.invoke(final_prompts)
print(response.content)
```

```python
#Chaining - Runnable
response = llm.invoke(dynamic_format_prompts.invoke({'language':'Hindi', 'query':'I Love You'}))
print()response.content
```

```python
from langchain_core.output_parsers import StrOutputParser
out = StrOutputParser() #Used so that we can avoid res.content each time

#Creating a method to add in chain to upper case the o/p
def transform_case(result:str):
    return result.upper()

#Chaining - Runnable
chains = prompts | llm | out | transform_case
response = chains.invoke({'language':'Hindi', 'query':'I Love You'})
#print(response.content)
print(response)
```
