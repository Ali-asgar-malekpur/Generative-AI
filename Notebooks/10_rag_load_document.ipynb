```python
# pip install pypdf
#Text Document Loader
from langchain_community.document_loaders import TextLoader
loader = TextLoader('../data/speech.txt')
print(loader) # Will output an Object of textloader

texts = loader.load() # Output a list always
#List contains: metadata and page_content
print(len(texts))

first_page = texts[0]
print(first_page.page_content)
```

```python
#PDF Document Loader
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader('../data/medical_report.pdf')
docs = loader.load()
print(len(docs)) #O/P: 9
#Since the pages in PDF is 9, it will give a list of 9 in length
first_page = docs[0]
print(first_page)
```

```python
#CSV Document Loader
from langchain_community.document_loaders import CSVLoader
loader = CSVLoader('../data/employee_records.csv')
docs = loader.load()
print(len(docs)) #O/P: 30000
#Since the pages in PDF is 30000, it will give a list of 30000 in length
content_csv = docs[:5]
print(content_csv)
#O/P: Metadata & acutal content for each list element
```

```python
#Web Page Loader (also requires beautifulsoup4 pkg)
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader(web_path='https://techsimplus.com/')
docs = loader.load()
print(docs[0].metadata)
print(docs[0].page_content)
#O/P: Metadata & acutal content for each list element
```
