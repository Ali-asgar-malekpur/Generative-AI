```python
documents = [
    'Machine learning is a subset of artificial intelligence',
    'Deep learning uses neutral networks with multiple layers',
    'Python is a programming language popular for data science',
    'India won the cricket world cup in 2011',
    'AI models can understand and generate human language'
]
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings_model = OpenAIEmbeddings(model = 'text-embedding-3-large')
vector = embeddings_model.embed_documents(documents)
print(len(vector)) #O/p: 5
print(len(vector[0])) #3072 vectors

# These vectors are stored in vector db
```

```python
from langchain_chroma import Chroma

vector_store = Chroma.from_texts(
    texts = documents,
    embeddings=embeddings_model,
    persist_directory='./vector_db'
)

query = 'Machine Learnings'
result = vector_store.similarity_search(query, k=2) #k for number of results in list
print(result)
```
