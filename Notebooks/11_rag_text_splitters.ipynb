```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
texts = 'Today, i want to speak from the heart about the threee splitters'\
'that explain the future of our nation: our youth, our education system and our farmers'
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 50, chunk_overlap = 20)
#chunk_size: No of Characters
#chunk_overlap: for the 2nd chunk and so on, it will pickup the 20 character from
#it's previous chunk to relate the data
```

```python
chunks = splitter.split_text(texts)
print(chunks)
#O/P: ['Today i want to speak from',
# 'from the heart about the threee', 
# 'threee splitters that explain the future'], etc
```

```python
from langchain_community.document_loaders import TextLoader, PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_loader = TextLoader('../data/speech.txt')
full_text = text_loader.load()

pdf_loader = PyPDFLoader('../data/medical_report.pdf')
full_pdf_text = pdf_loader.load()

splitter = RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 20)
text_chunks = splitter.split_documents(full_text)
len(text_chunks)

pdf_chunks = splitter.split_documents(full_pdf_text)
len(pdf_chunks)
```
